## Overview

This notebook evaluates all combinations of COVID-19 data variants against a set of major economic indicators across four pandemic phases. Each combination is scored by `mathx.score_x_vs_y_dict()`, a weighted composite synthesizing five analytical layers — Granger causality (0.30), VAR/VECM dynamics (0.25), Engle-Granger cointegration (0.20), lag correlation (0.15), and contemporaneous correlation (0.10) — into a single comparable score in [0, 100]. The analysis is run twice: once on levels (Part 1) and once on percentage changes (Part 2).

---

## Scoring Framework

| Component | Weight | Measures |
|---|---|---|
| Granger Causality | 0.30 | Causal direction and significance (X→Y vs. Y→X) |
| VAR/VECM | 0.25 | IRF peak effect, FEVD x-share, model stability |
| Cointegration | 0.20 | Long-run equilibrium relationship |
| Lag Correlation | 0.15 | Peak delayed association and X-leads-Y directionality |
| Correlation | 0.10 | Contemporaneous association strength |

---

## Pandemic Phase Definitions

| Phase | Period | Characterization |
|---|---|---|
| **Full Period** | 01/22/2020 – 09/14/2022 | Entire dataset spanning all phases |
| **Initial Shock** | 01/22/2020 – 06/07/2020 | Pandemic emergence, market crash, initial policy response |
| **Adaptation** | 06/08/2020 – 11/25/2021 | Vaccine development and rollout, market recovery, Delta wave |
| **Recovery** | 11/26/2021 – 09/14/2022 | Omicron wave, Fed tightening, post-pandemic normalization |

## Summary and Conclusions

### Part 1: Levels

The levels analysis scores peak sharply during the Initial Shock phase and decline steadily through each subsequent period. The top composite scores by phase are: Initial Shock **66.06**, Adaptation **54.75**, Full Period **48.74**, and Recovery **51.99**. The Initial Shock score is by far the highest across both parts and all phases, confirming that COVID data exerted its strongest causal influence on economic indicators during the acute crash period, before markets had time to adapt or price in the pandemic as a persistent background condition.

`covid_deaths` is the dominant X series across the Full Period, Adaptation, and Recovery phases, appearing in 7 of the top 10 rows in each. This reflects a shift in the market's informational focus from raw case volume to fatality outcomes as the pandemic matured. In the Initial Shock phase, `covid_cases` and `covid_c&d` (combined) dominate instead, consistent with investors responding to the breadth of infection spread during the initial crash rather than its severity.

On the Y side, `agr1` leads in the Full Period, `stk_mkt` and `intrt_rts` lead in the Initial Shock phase, `intrt_rts` leads in Adaptation, and `metals` dominates in Recovery with six of the top 10 positions. The rise of `metals` in the Recovery phase likely reflects commodity market dynamics driven by supply chain disruption and inflation expectations rather than direct pandemic causality — consistent with the weakening of the COVID signal by that period.

The cumulative (`cuml. x = True`) and rolling average (`roll. x = True`) transformations appear inconsistently across phases. Neither transformation dominates in the levels analysis, suggesting that the raw COVID series carries the primary causal signal at this data frequency.

### Part 2: Percentage Changes

The percentage-change analysis produces the highest scores of any combination across the entire analysis. The top scores by phase are: Initial Shock **60.98**, Adaptation **56.60**, Recovery **55.67**, and Full Period **51.81**. Scores decline progressively from Initial Shock through Adaptation and Recovery — the same directional pattern as Part 1 — with the Full Period score falling lowest of all four. This confirms that COVID's short-run causal influence on economic indicator percentage changes was most concentrated in the crash period and attenuated steadily thereafter.

A key structural difference from Part 1 is that `pct. x = True` and `pct. y = True` are set in every top-10 row across all four phases in Part 2 by design. Within that constraint, both `cuml. x` and `roll. x` appear frequently as `True`, indicating that cumulative and rolling transformations add incremental signal in percentage-change form — the reverse of their marginal contribution in Part 1.

`covid_c&d` and `covid_cases` share dominance as the X series in the Initial Shock and Recovery phases, while `covid_deaths` leads in the Adaptation phase and ties with `covid_c&d` in the Full Period top row. On the Y side, `stk_mkt` leads in the Initial Shock phase, `metals` leads in Adaptation and Recovery, and `exchg_rts` (exchange rates) dominates the Full Period — a notable finding suggesting that currency markets maintained a more persistent short-run relationship with COVID data across the entire dataset than equity or commodity markets did.

### Cross-Phase Score Progression

The score progression across phases tells a consistent story in both parts: causality was sharpest and most concentrated at the onset of the pandemic and weakened as the crisis evolved. The Initial Shock phase produces the highest top scores in both Part 1 (66.06) and Part 2 (60.98). Scores then decline through Adaptation and compress further by Recovery, with less differentiation between the top 10 candidates — meaning no single pairing dominated clearly by the later phases. This compression is consistent with COVID transitioning from the primary market driver to one factor among many as the economy adapted.

The Full Period score is the lowest of all four periods in both parts, which is initially counterintuitive but reflects the averaging effect of pooling high-signal (Initial Shock) and low-signal (Adaptation, Recovery) data: the composite relationship across the full 2.5-year window is weaker than in any individual phase because the causal structure changed substantially over time.

### Optimal Candidates (Table 2.5)

The top-ranked candidate for each phase, selected across both Part 1 and Part 2, is summarized in Table 2.5:

| Phase | cuml. x | roll. x | pct. x | x | pct. y | y | Score |
|---|---|---|---|---|---|---|---|
| **Full Period** | False | False | True | covid_deaths | True | exchg_rts | 51.81 |
| **Initial Shock** | True | True | False | covid_c&d | False | stk_mkt | 66.06 |
| **Adaptation** | True | True | True | covid_deaths | True | metals | 56.60 |
| **Recovery** | True | True | True | covid_c&d | True | metals | 55.67 |

The Initial Shock candidate (`covid_c&d` → `stk_mkt`, levels with cumulative and rolling transformations, score **66.06**) is the strongest single pairing in the dataset and the recommended input for any downstream analysis targeting the maximum-signal phase. The Full Period candidate (`covid_deaths` → `exchg_rts`, percentage changes, score **51.81**) is the appropriate choice for analyses requiring a relationship that holds across the entire 2020–2022 window.
